#### Starter Code for Question 1

```python
from collections import deque
from typing import List

# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

#converting tree to list
def tree_to_list_bfs(root: TreeNode):
    if not root:
        return []
    result = []
    queue = deque([root])
    while queue:
        node = queue.popleft()
        result.append(node.val)
        if node.left:
            queue.append(node.left)
        if node.right:
            queue.append(node.right)
    return result

#checking nearest duplicate in list 
def first_duplicate(nums: List[int]) -> int: 
    checked = set()
    for n in nums:
        if n in checked:
            return n
        checked.add(n)
    return -1

#combining the two to get desired result
def is_duplicate(root: TreeNode) -> int:
    values = tree_to_list_bfs(root)
    return first_duplicate(values)
```

```python
#test tree
root = TreeNode(6)
root.left = TreeNode(5)
root.right = TreeNode(7)
root.left.left = TreeNode(3)
root.left.right = TreeNode(9)  # duplicate of 3

# Test the function
print(is_duplicate(root))  # Expected output: 1 (duplicate exists)
```

However, this can also be achieved by swapping the elements in place instead of using 2 for loops, which I believe is a cleaner and concise code. Code as follows:

```python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    insert_pos = 0
    for i in range(len(nums)):
        if nums[i] != 0:
            #moving non-zero elements to the front of the list by swapping them
            nums[insert_pos], nums[i] = nums[i], nums[insert_pos]
            insert_pos += 1
    return nums
```
